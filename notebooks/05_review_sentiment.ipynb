# Phase 5: Review Sentiment Analysis (NLP)

**Business question:** What are unhappy customers actually complaining about, and does
a negative experience stop them from coming back?

**Approach** (`src/sentiment.py`)
- Reviews are in Portuguese, so a **multilingual BERT** model fine-tuned on product
  reviews (`nlptown/bert-base-multilingual-uncased-sentiment`) predicts 1–5 stars from the text.
- **Validation:** compare the model with the stars customers actually gave.
- **Complaint themes:** a Portuguese keyword dictionary tags late delivery, wrong items,
  damage, quality and seller-support issues.
- **Link to churn:** repeat-purchase rate by the sentiment of a customer's review.

> The first run downloads the model (~650 MB). Scoring all ~40K reviews takes a while on a
> laptop CPU (Apple Silicon GPUs are used automatically), so start with `limit=5000`.
