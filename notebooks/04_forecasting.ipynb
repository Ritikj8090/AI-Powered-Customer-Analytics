# Phase 4: Revenue Forecasting (Prophet)

**Business question:** How much revenue should each category expect over the next
quarter, so inventory and marketing can plan ahead?

**Approach** (`src/forecasting.py`)
- Weekly revenue for the whole business and the top 8 categories.
- **Prophet** with a Black Friday holiday effect (the biggest spike in the data). Yearly
  seasonality is off because there are fewer than two years of history to learn it from.
- **Backtest:** hide the last 12 weeks, forecast them, and compare with a naive baseline
  (average of the 4 weeks before). A forecast is only useful if it beats the naive one.
- **Metrics:** MAPE (average % error per week) and WAPE (total error ÷ total revenue).
  WAPE is the fairer number because MAPE blows up on low-revenue weeks.
